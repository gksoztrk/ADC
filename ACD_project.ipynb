```python
import os
import shutil

path = r"C:\Users\InesNeves\.cache\kagglehub\datasets\yelp-dataset\yelp-dataset\versions\4"
files = os.listdir(path)
print("Files in dataset:", files)

target_path = r"C:\Users\InesNeves\Desktop\ACD\Project"

# Create the destination directory if it does not exist
os.makedirs(target_path, exist_ok=True)

# Loop through and move each file/folder
for file_name in os.listdir(path):
    src_file = os.path.join(path, file_name)
    dst_file = os.path.join(target_path, file_name)

    shutil.move(src_file, dst_file)
    print(f"Moved: {file_name}")

print("\nAll files moved successfully!")
```

```python
import pandas as pd

# Get all .json files in the folder
json_files = [f for f in os.listdir(target_path) if f.endswith(".json")]

for file_name in json_files:
    file_path = os.path.join(target_path, file_name)

    print("=" * 70)
    print(f" FILE: {file_name}")
    print("=" * 70)

    # Read 5 rows safely without loading the full file into memory
    df = pd.read_json(file_path, lines=True, nrows=5)

    print("\n--- Schema Info ---")
    df.info()

    print("\n--- Preview (First 5 Rows) ---")
    display(df.head())
    print("\n")
```
